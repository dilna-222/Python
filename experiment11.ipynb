```python
#generate positive list of integers from a given list of integers
numbers=list(map(int,input("Enter the numbers:").split()))
print("The numbers:",numbers)
result=[i for i in numbers if i>0]
print("Positive numbers are:",result)
```

```python
#square of N numbers
n=int(input("Enter the limit:"))
print("The limit is:",n)
result=[i*i for i in range(1,n+1)]
print("Squares:",result)
```

```python
#form a list of vowels selected from a given word
word=input("Enter a word:")
print("The word is:",word)
result=[i for i in word if i.lower() in "aeiou"]
print("Vowels:",result)
```
