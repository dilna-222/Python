```python
#sort dictionary in ascending and descensing order of their roll number
roll=input("Enter Roll no(Space seperated): ").split()
name=input("Enter Name(Space seperated): ").split()
mydict=dict(zip(roll,name))
print("Original dictionary is",mydict)
print("Ascending order:",dict(sorted(mydict.items())))
print("Descending order:",dict(sorted(mydict.items(),reverse=True)))
```

```python
#merge two dictionaries and display it
dic1=dict(zip(input("Enter 3 roll number:").split(),input("Enter 3 names:").split()))
dic2=dict(zip(input("Enter 2 roll number:").split(),input("Enter 2 names:").split()))
print("dictionary1=",dic1)
print("dictionary2=",dic2)
dic1.update(dic2)
print("Merged dictionary:",dic1)
```
