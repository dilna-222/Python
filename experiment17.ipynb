```python
#read a file line by line and store it into a list
f=open("sample.txt","r")
lines=[]
for line in f:
    lines.append(line)
f.close()
print("List of lines: ",lines)
```

```python
#Copy odd lines of one file to another
f1=open("sample.txt","r")
f2=open("odd.txt","w")

lines= f1.readlines()

for i in range(len(lines)):
    if i%2==0:
        f2.write(lines[i])
f1.close()
f2.close()

print("Odd lines copied successfully")
    
```

```python
#python program to write a ditionary to a csv file, then read and display its contents

import csv
#dictionary
data=[
    {'Name': 'Anu', 'Age':20, 'City':'Kochi'},
    {'Name': 'Rahul', 'Age':22, 'City':'Trivandrum'},
    {'Name': 'Meera', 'Age':21, 'City':'Kozhikode'},
]

#Write dictionary to csv file
f=open("student.csv", "w", newline="")

writer=csv.DictWriter(f, fieldnames=['Name', 'Age', 'City'])

writer.writeheader()
writer.writerows(data)

f.close()

print("Dictionary written to CSV file successfully")

#Read the CSV file
f=open("student.csv", "r")

reader=csv.DictReader(f)

print("\nContents of CSV file:")

for row in reader:
    print(dict(row))

f.close()
```
