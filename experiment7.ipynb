```python
#print out all colors from colorset 1 not contained in colorset 2
set1=set(input("Enter colors in set1:").split())
set2=set(input("Enter colors in set2:").split())
print("Set1=",set1)
print("Set2=",set2)
print("The colors in set1 but not in set2:",set1.difference(set2))
```

```python
#printout all colors from colorset1 and colorset2
s1=set(input("Enter colors in set1:").split())
s2=set(input("Enter colors in set2:").split())
print("Set1=",s1)
print("Set2=",s2)
print("The colors in both sets are:",s1.union(s2))
```
