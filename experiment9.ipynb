```python
#add "ing" to the end of a string.if it already ends with "ing" use "ly".
s=input("Enter a string: ")
if s[-3:] == "ing":
    print(s + "ly")
else:
    print(s +"ing")
```

```python
s=input("Enter a string: ")

if s == s[::-1]:
    print("Palindrome")
else:
    print("Not a palindrome")
```
