```python
#change a given string to a new string where the first and last chars have been exchanged
s=input("Enter a string: ")
new_string = s[-1] + s[1:-1] + s[0]
print("New string:", new_string)
```

```python
#get a string made of first 2 and last 2 chars from a given string
s=input("Enter a string:")
new_string = s[:2] + s[-2:]
print("New string:", new_string)
```

```python
#create a single string seperated with space from two strings by swapping the character at position 1.
s1=input("Enter first string:")
s2=input("Enter second string:")

new_s1 = s2[0] + s1[1:]
new_s2 = s1[0] + s2[1:]

print(new_s1 + " " + new_s2)
```

```python
#get a string from a given string where all occurences of its first char have been changed to $,except the first char itself
s=input("Enter a string: ")
new_string = s[0] + s[1:].replace(s[0],'$')
print("New string:", new_string)
```
